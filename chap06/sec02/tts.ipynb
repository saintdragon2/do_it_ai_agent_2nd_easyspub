```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')# 환경 변수에서 API 키 가져오기
client = OpenAI(api_key=OPENAI_API_KEY) # 오픈AI 클라이언트의 인스턴스 생성
```

```python
# 출처: https://developers.openai.com/api/docs/guides/text-to-speech

speech_file_path =  "./speech.mp3"

with client.audio.speech.with_streaming_response.create(
    model="gpt-4o-mini-tts",
    voice="alloy",
    input="Today is a wonderful day to build something people love!",
    instructions="Speak in a cheerful and positive tone.",
) as response:
    response.stream_to_file(speech_file_path)

import IPython.display as ipd
ipd.Audio(speech_file_path)
```

```python
voice = "ash"
speech_file_path =  f"./hello_world_{voice}.mp3"

with client.audio.speech.with_streaming_response.create(
    model="gpt-4o-mini-tts",
    voice=voice,
    input=f"Hello World! I'm {voice}. This is a TTS test.",
    instructions="토익 듣기평가 문제에 나오는 스타일로 말해줘.",
) as response:
    response.stream_to_file(speech_file_path)

import IPython.display as ipd
ipd.Audio(speech_file_path)
```

```python
import json

with open('../images/image_quiz_eng.json', 'r', encoding='utf-8') as f:
    eng_dict = json.load(f)

eng_dict
```

```python
voices = ['alloy', 'ash', 'coral', 'echo', 'fable', 'onyx', 'nova', 'sage' , 'shimmer']
for q in eng_dict:
    no = q['no']
    quiz = q['eng']                             # (1) 시작
    quiz = quiz.replace("- (1)", "- One.\t")
    quiz = quiz.replace("- (2)", "- Two.\t")
    quiz = quiz.replace("- (3)", "- Three.\t")
    quiz = quiz.replace("- (4)", "- Four.\t")   # (2)

    print(no, quiz)
    voice = voices[no % len(voices)] # (2)

    with client.audio.speech.with_streaming_response.create(
        model="gpt-4o-mini-tts",
        voice=voice,
        input=quiz,
        instructions="토익 듣기평가 문제에 나오는 스타일로 말해줘.",
    ) as response:
        response.stream_to_file(f"../audio/quiz_{no}_{voice}.mp3")
```
