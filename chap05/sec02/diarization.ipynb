```python
from openai import OpenAI
from dotenv import load_dotenv
import os

load_dotenv()
OPENAI_API_KEY = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=OPENAI_API_KEY)
```

```python
import base64

def to_data_url_mp3(path: str) -> str:
    with open(path, "rb") as fh:
        return "data:audio/mpeg;base64," + base64.b64encode(fh.read()).decode("utf-8")

with open("../audio/싼기타_비싼기타.mp3", "rb") as audio_file:
    transcript = client.audio.transcriptions.create(
        model="gpt-4o-transcribe-diarize",
        file=audio_file,
        response_format="diarized_json",
        chunking_strategy="auto",
        extra_body={
            "known_speaker_names": ["이성용"],
            "known_speaker_references": [
                to_data_url_mp3("../audio/lsy_audio_2023_first8s.mp3")
            ],
        },
    )
```

```python
for segment in transcript.segments:
    print(
        segment.speaker, 
        segment.text, 
        round(segment.start, 2), 
        round(segment.end, 2)
    )
```

```python
import pandas as pd

df = pd.DataFrame(
    [
        {
            "speaker": segment.speaker,
            "text": segment.text,
            "start": round(segment.start, 2),
            "end": round(segment.end, 2),
        }
        for segment in transcript.segments
    ]
)

display(df.head(10))

df.to_excel("./싼기타_비싼기타.xlsx", index=False)
```
