## 05-3 회의록을 요약 정리하기

앞서 만든 녹취록을 정리해서 회의록으로 만들어봅시다. 

### Do It! 실습: 전체 회의 내용 요약하기
앞서 만들었던 CSV 파일에서 '이성용', 'A'로 표시되었던 부분 중, 'A'로 표시되었던 것을 'AI'로 수정하고 GPT에게 전체 회의 내용 요약을 요청해보겠습니다. 

1. 단계별로 진행하기 위해, 주피터 노트북을 생성합니다.
2. 그리고 앞 절에서 만들었던 엑셀 파일을 열어 판다스 데이터프레임으로 읽습니다. 전에도 이야기 했지만, pandas는 파이썬에서 데이터분석을 할 때 많이 쓰는 패키지로써 엑셀 같은 기능을 한다고 보면 됩니다. (1) 파일을 읽고, (2) 잘 읽어왔는지 `.head(10)`으로 10행까지 디스플레이합니다. `display`는 주피터 노트북에서 데이터프레임을 볼 때 많이 쓰는 함수입니다. 

```python
import pandas as pd

df_meeting = pd.read_excel("../sec02/싼기타_비싼기타.xlsx")
display(df_meeting.head(10))
```

3. "A" 라고 표현된 부분을 "AI"라고 수정합니다.

|판다스가 익숙하지 않은 독자들을 위해 `.apply`에 대해 잠시 설명하겠습니다. `.apply는 판다스 데이터프레임에서 많이 쓰는 메소드(함수)입니다. 이 코드는 "speaker" 열의 값을 하나씩 읽어서 x라 하고 그 x가 "A"라고 되어 있을 때는 'AI'로 교체하고, 그렇지 않을 때는 x 그대로 쓰라"는 의미입니다. |
|--|

```python
df_meeting["speaker"] = df_meeting["speaker"].apply(lambda x: "AI" if x == "A" else x)
df_meeting.head(10)
```

4. 회의록 요약을 위해서는 speaker와 text 열만 필요하니까, 해당 열만 추출해서 text 로 변환해봅시다. 빈 텍스트 `meeting_str`을 만들고, 한 행씩 읽어서 "화자: 내용"의 패턴으로 정리했습니다. `.iterrows()`는 한 줄씩 읽는 판다스 데이터프레임의 메소드(함수)입니다. 

```python
meeting_str = ""
for index, row in df_meeting.iterrows():
    meeting_str += f"{row['speaker']}: {row['text']}\n"

print(meeting_str)
```

5. GPT에게 어떻게 회의록을 써야 하는지 지시하는 프롬프트를 작성합니다.




```python
system_prompt = f"""
너는 회의 내용을 요약하는 봇이다. 아래 회의록을 읽고, 주요 내용을 요약하라.
결과는 마크다운 형식으로 작성한다.
아래 형식에 맞추어 작성하라.
# 회의 제목

## 주요 내용

## 참석자별 입장

## 결정 사항
"""
```

6. GPT에게 요청하기 전에 오픈AI의 API client 를 선언합니다.
* GPT를 활용할 때는 보안 문제에 유의해야 합니다. 유출되면 안 되는 기밀 정보가 있다면 로컬 모델을 사용하는 것을 권장합니다.

```python
from openai import OpenAI
from dotenv import load_dotenv
import os
load_dotenv()
api_key = os.getenv('OPENAI_API_KEY')
client = OpenAI(api_key=api_key)
```

7. 이제 GPT 클라이언트를 이용해 프롬프트를 보냅니다. 이 요청에 대한 답신은 `response`에 담았습니다. 

```python
response = client.responses.create(
    model="gpt-5.6-luna",
    instructions=system_prompt,
    input=meeting_str,
    reasoning={
        "effort": "none" # 'none', 'low', 'medium', 'high', 'xhigh'
    }
)

print(response)
```

불필요한 내용이 많으니까, output_text 만 출력해봅시다. 

```python
print(response.output_text)
```

8. 회의록을 `guitar_summary.md` 로 저장합니다. 내용은 출력된 결과와 동일합니다. 

```python
with open('./guitar_summary.md', 'w', encoding='utf-8') as f:
    f.write(response.output_text)
```
