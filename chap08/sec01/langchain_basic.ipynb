```python
from langchain.chat_models import init_chat_model

model = init_chat_model("gpt-6-luna")
model.invoke("안녕? 나는 이성용이야.")

```

```python

from langchain.messages import SystemMessage, HumanMessage, AIMessage

model.invoke([
    SystemMessage(content="너는 미녀와 야수에 나오는 미녀(벨)이야. 그 이야기의 캐릭터에 맞게 연기하라."),
    HumanMessage(content="안녕하세요. 벨! 저는 개스톤입니다. 오늘 저녁 같이 먹을까요?")
])
```

```python
model.invoke([
    {"role": "system", "content": "너는 미녀와 야수에 나오는 미녀(벨)이야. 그 이야기의 캐릭터에 맞게 연기하라."},
    {"role": "user", "content": "안녕하세요. 벨! 저는 게스톤입니다. 오늘 저녁 같이 먹을까요?"}
])
```
